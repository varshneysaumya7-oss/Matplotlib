```python
## Q11.  Highlight the lowest bar&#x20;
#Draw a bar chart of avg_marks. Make the bar with the lowest average red and all other bars blue.
#(a) Make a list of colors, one for each bar.
#(b) Add a title and axis labels.


import matplotlib.pyplot as plt

# Branch and average marks
branch = ['CSE', 'ECE', 'ME', 'CE', 'IT']
avg_marks = [85, 78, 72, 80, 90]

# Find the lowest average marks
lowest = min(avg_marks)

# Create color list
colors = ['red' if marks == lowest else 'blue' for marks in avg_marks]

# Draw bar chart
plt.bar(branch, avg_marks, color=colors)

# Add title and axis labels
plt.title("Average Marks by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Marks")

# Display chart
plt.show()
```

```python
#Q13.  Histogram with a mean line
#Draw a histogram of marks and mark the average on it.
#(a)	Use 8 bins.
#(b)	Draw a red dashed vertical line at the mean with plt.axvline and add a legend.
#(c)	Add a title and axis labels.


import matplotlib.pyplot as plt
import numpy as np

# Marks of 24 students
marks = [85, 78, 90, 65, 72, 88, 95, 76,
         82, 69, 91, 74, 80, 87, 93, 68,
         77, 84, 89, 71, 96, 79, 86, 73]

# Calculate mean
mean_marks = np.mean(marks)

# Draw histogram with 8 bins
plt.hist(marks, bins=8, color='skyblue', edgecolor='black',
         label='Student Marks')

# Draw red dashed vertical line at mean
plt.axvline(mean_marks, color='red', linestyle='--',
            label='Mean Marks')

# Add legend
plt.legend()

# Add title and axis labels
plt.title("Histogram of Marks with Mean")
plt.xlabel("Marks")
plt.ylabel("Frequency")

# Display graph
plt.show()
```

```python
#Q16.  Scatter plot with style
#Draw a scatter plot of attendance (x) against marks (y).
#(a)	Use red dots of size 60 and alpha 0.7.
#(b)	Add a grid, a title and axis labels.

import matplotlib.pyplot as plt

plt.scatter(students['attendance'], students['marks'], color='red', s=60, alpha=0.7)
plt.grid(True)
plt.title("Attendance vs Marks")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
#Q17.  Average marks per branch
#Calculate the average marks of each branch from the students table and show them as a bar chart.
#(a)	Use groupby to find the average marks of each branch.
#(b)	Draw a bar chart of the result, with the value written above each bar.
#(c)	Add a title and axis labels.

import matplotlib.pyplot as plt

avg_marks = students.groupby('branch')['marks'].mean()

plt.bar(avg_marks.index, avg_marks.values)

for i, v in enumerate(avg_marks.values):
    plt.text(i, v, round(v, 2), ha='center')

plt.title("Average Marks per Branch")
plt.xlabel("Branch")
plt.ylabel("Average Marks")

plt.show()
```

```python
## Q18.  Students in each branch&#x20;
#Count how many students are in each branch and show the share of each branch as a pie chart.
#(a)   Use value_counts to count the students of each branch.
#(b) Draw the pie chart with branch names and percentages, and add a title.
#(c) Pull out the largest slice a little using explode.

import matplotlib.pyplot as plt

branch_count = students['branch'].value_counts()

explode = [0.1 if i == 0 else 0 for i in range(len(branch_count))]

plt.pie(branch_count, labels=branch_count.index, autopct='%1.1f%%', explode=explode)

plt.title("Students in Each Branch")

plt.show()
```

```python
#Q19.  Growth in sales
#Find how much sales grew from 2025 to 2026 in each month and show it as a bar chart.
#(a)	Make a list growth = sales_2026 minus sales_2025 for each month.
#(b)	Draw the bar chart and make the bar of the month with the biggest growth red. Other bars are blue.
#(c)	Add a title, axis labels and the value above each bar.

import matplotlib.pyplot as plt

growth = [s26 - s25 for s26, s25 in zip(sales_2026, sales_2025)]

colors = ['red' if v == max(growth) else 'blue' for v in growth]

plt.bar(months, growth, color=colors)

for i, v in enumerate(growth):
    plt.text(i, v + 0.5, v, ha='center')

plt.title("Monthly Sales Growth (2025-2026)")
plt.xlabel("Months")
plt.ylabel("Sales Growth")
plt.show()
```

```python
#Q20.  Marks report
#Make a report graph of the marks of all students and save it as an image.
#(a)	Draw a histogram of marks (8 bins). Add a red dashed line for the mean and a green dotted line for the median, with a legend.
#(b)	Add a title and axis labels, and print the mean and median.
#(c)	Save the figure as marks_report.png with dpi = 200 before plt.show().

import matplotlib.pyplot as plt
import numpy as np

# Marks of all 24 students from Dataset 2
marks = students['marks']

# Calculate mean and median
mean_marks = np.mean(marks)
median_marks = np.median(marks)

# Print mean and median
print(f"Mean Marks: {mean_marks:.2f}")
print(f"Median Marks: {median_marks:.2f}")

# (a) Draw histogram with 8 bins, mean line, and median line
plt.hist(marks, bins=8, color='skyblue', edgecolor='black', label='Student Marks')
plt.axvline(mean_marks, color='red', linestyle='--', label=f'Mean ({mean_marks:.2f})')
plt.axvline(median_marks, color='green', linestyle=':', label=f'Median ({median_marks:.2f})')

# (b) Add legend, title, and axis labels
plt.legend()
plt.title("Marks Report: Distribution, Mean & Median")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

# (c) Save figure before plt.show()
plt.savefig("marks_report.png", dpi=200, bbox_inches='tight')
plt.show()
```
